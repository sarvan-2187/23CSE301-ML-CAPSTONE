```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Phishing_Email.csv')
raw_columns = list(df.columns)

print(f"Shape: {df.shape[0]:,} rows x {df.shape[1]} columns")
print()
print("Columns:", list(df.columns))
print()
print("Dtypes:")
print(df.dtypes)
print()
df.head(3)
```

```python
class_counts = df['Email Type'].value_counts()
class_pct = df['Email Type'].value_counts(normalize=True) * 100

summary = pd.DataFrame({'count': class_counts, 'percent': class_pct.round(2)})
summary
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

sns.barplot(x=class_counts.index, y=class_counts.values, ax=axes[0],
            hue=class_counts.index, palette=['#4C72B0', '#DD8452'], legend=False)
axes[0].set_title('Email Count by Class')
axes[0].set_xlabel('Email Type')
axes[0].set_ylabel('Number of Emails')
for i, v in enumerate(class_counts.values):
    axes[0].text(i, v + 150, f'{v:,}', ha='center', fontweight='bold')

axes[1].pie(class_counts.values, labels=class_counts.index, autopct='%1.1f%%',
            colors=['#4C72B0', '#DD8452'], startangle=90)
axes[1].set_title('Class Proportion')

plt.tight_layout()
plt.show()
```
